```python
# 1. Identify the daily consumption columns
date_cols = [
    col for col in df_clean.columns
    if col not in ["CONS_NO", "FLAG", "missing_percentage"]
]

# Sort columns chronologically
date_cols = sorted(
    date_cols,
    key=lambda col: pd.to_datetime(col, format="%Y/%m/%d")
)

dates = pd.to_datetime(date_cols, format="%Y/%m/%d")
```

```python
# Extract consumption values
X = df_clean[date_cols].astype(float)

# Convert to NumPy for calculations
arr = X.to_numpy(dtype=float)

# 2. Missing-data quality: valid days
valid_days = X.notna().sum(axis=1)

# 3. Zero-consumption behavior
observed_zero_rate = (
    X.eq(0).sum(axis=1) / valid_days.replace(0, np.nan)
)
```

```python
# 4. Daily changes: only consecutive calendar days
date_gaps = np.diff(dates.values).astype("timedelta64[D]").astype(int)

# A pair is calendar-consecutive only when the date gap is one day
consecutive_pairs = date_gaps == 1

valid_pairs = (
    np.isfinite(arr[:, :-1])
    & np.isfinite(arr[:, 1:])
    & consecutive_pairs[np.newaxis, :]
)

daily_changes = np.abs(arr[:, 1:] - arr[:, :-1])
daily_changes[~valid_pairs] = np.nan

valid_change_days = valid_pairs.sum(axis=1)
```

```python
def row_nanmean(values):
    with np.errstate(invalid="ignore"):
        return np.nanmean(values, axis=1)

def row_nanstd(values):
    with np.errstate(invalid="ignore"):
        return np.nanstd(values, axis=1, ddof=1)

avg_daily_change = row_nanmean(daily_changes)
std_daily_change = row_nanstd(daily_changes)
```

```python
# Percentile calculated row by row, ignoring NaNs
p95_daily_change = np.array([
    np.nanpercentile(row, 95) if np.isfinite(row).any() else np.nan
    for row in daily_changes
])

# 5. Trend slope: linear trend against actual dates
# Center dates to avoid large numeric values
time_days = (dates - dates[0]).days.to_numpy(dtype=float)
trend_slope = np.full(len(df_clean), np.nan)

for i, row in enumerate(arr):
    valid = np.isfinite(row)

    if valid.sum() >= 2:
        trend_slope[i] = np.polyfit(
            time_days[valid], row[valid], 1
        )[0]
```

```python
# 6. Early-to-late consumption change
# Compare first and last 20% of the calendar period
n_dates = len(date_cols)
period_size = max(1, int(np.ceil(n_dates * 0.20)))

early_mean = X.iloc[:, :period_size].mean(axis=1)
late_mean = X.iloc[:, -period_size:].mean(axis=1)

early_late_change = late_mean - early_mean
```

```python
# 7. Monthly mean variability
monthly_means = pd.DataFrame(index=df_clean.index)

for month in dates.to_period("M").unique():
    month_cols = [
        col for col, date in zip(date_cols, dates)
        if date.to_period("M") == month
    ]

    # Require at least 15 observed days in the month
    month_data = X[month_cols]
    month_mean = month_data.mean(axis=1)
    enough_days = month_data.notna().sum(axis=1) >= 15

    monthly_means[str(month)] = month_mean.where(enough_days)

monthly_mean_std = monthly_means.std(axis=1, ddof=1)
```

```python

features_now = pd.DataFrame({
    "valid_days": valid_days,
    "observed_zero_rate": observed_zero_rate,
    "valid_change_days": valid_change_days,
    "avg_daily_change": avg_daily_change,
    "std_daily_change": std_daily_change,
    "p95_daily_change": p95_daily_change,
    "trend_slope": trend_slope,
    "early_late_change": early_late_change,
    "monthly_mean_std": monthly_mean_std
})
```

```python
n_dates = len(date_cols)
period_size = max(1, int(np.ceil(n_dates * 0.20)))

early_data = X.iloc[:, :period_size]
late_data = X.iloc[:, -period_size:]

early_valid_days = early_data.notna().sum(axis=1)
late_valid_days = late_data.notna().sum(axis=1)

early_median = early_data.median(axis=1)
late_median = late_data.median(axis=1)

enough_data = (
    (early_valid_days >= 15) &
    (late_valid_days >= 15)
)

early_late_change = (
    late_median - early_median
).where(enough_data)

features_now["early_late_change"] = early_late_change.to_numpy()
```

```python
missing = X.isna().to_numpy()
n_consumers, n_days = missing.shape

missing_percentage = missing.mean(axis=1) * 100
max_missing_streak = np.zeros(n_consumers, dtype=int)
missing_block_count = np.zeros(n_consumers, dtype=int)
isolated_missing_count = np.zeros(n_consumers, dtype=int)
leading_missing_days = np.zeros(n_consumers, dtype=int)
trailing_missing_days = np.zeros(n_consumers, dtype=int)

for i, row in enumerate(missing):
    padded = np.r_[False, row, False]
    changes = np.diff(padded.astype(int))

    starts = np.where(changes == 1)[0]
    ends = np.where(changes == -1)[0]

    streak_lengths = ends - starts

    missing_block_count[i] = len(starts)

    if len(streak_lengths) > 0:
        max_missing_streak[i] = streak_lengths.max()

    if row[0]:
        leading_missing_days[i] = ends[0]

    if row[-1]:
        trailing_missing_days[i] = n_days - starts[-1]

    if n_days > 2:
        isolated_missing_count[i] = (
            row[1:-1]
            & ~row[:-2]
            & ~row[2:]
        ).sum()

missingness_summary = pd.DataFrame({
    "CONS_NO": df_clean["CONS_NO"].to_numpy(),
    "missing_percentage": missing_percentage,
    "max_missing_streak": max_missing_streak,
    "missing_block_count": missing_block_count,
    "isolated_missing_count": isolated_missing_count,
    "leading_missing_days": leading_missing_days,
    "trailing_missing_days": trailing_missing_days
})

missingness_summary.head()
```

```python


gap_categories = {
    "gaps_1_day": (1, 1),
    "gaps_2_3_days": (2, 3),
    "gaps_4_7_days": (4, 7),
    "gaps_8_30_days": (8, 30),
    "gaps_31_90_days": (31, 90),
    "gaps_over_90_days": (91, float("inf"))
}

gap_results = []

missing = X.isna().to_numpy()

for i, row in enumerate(missing):
    padded = np.r_[False, row, False]
    changes = np.diff(padded.astype(int))

    starts = np.where(changes == 1)[0]
    ends = np.where(changes == -1)[0]
    lengths = ends - starts

    result = {
        "CONS_NO": df_clean["CONS_NO"].iloc[i],
        "missing_percentage": missingness_summary["missing_percentage"].iloc[i],
        "total_missing_blocks": len(lengths),
        "max_missing_streak": lengths.max() if len(lengths) else 0
    }

    for name, (lower, upper) in gap_categories.items():
        result[name] = sum(
            lower <= length <= upper for length in lengths
        )

    gap_results.append(result)

gap_summary = pd.DataFrame(gap_results)

gap_summary.head()
```

```python
import numpy as np
import pandas as pd

dates = pd.to_datetime(date_cols, format="%Y/%m/%d")
X = X[date_cols].astype(float)
X_filled = X_filled[date_cols].astype(float)

features_compulsory = pd.DataFrame(index=df_clean.index)

# 1. Consumption statistics
features_compulsory["avg_consumption"] = X_filled.mean(axis=1)
features_compulsory["std_consumption"] = X_filled.std(axis=1)
features_compulsory["cv_consumption"] = (
    features_compulsory["std_consumption"] /
    features_compulsory["avg_consumption"].replace(0, np.nan)
)
features_compulsory["IQR"] = (
    X_filled.quantile(0.75, axis=1) -
    X_filled.quantile(0.25, axis=1)
)
features_compulsory["P10"] = X_filled.quantile(0.10, axis=1)
features_compulsory["P90"] = X_filled.quantile(0.90, axis=1)

# 2. Zero-consumption behavior from original observations
observed = X.notna().to_numpy()
arr = X.to_numpy()
zero = observed & (arr == 0)
nonzero = observed & (arr != 0)

features_compulsory["observed_zero_rate"] = (
    zero.sum(axis=1) /
    observed.sum(axis=1).clip(min=1)
)

longest_zero_streak = []
zero_episode_count = []
zero_transition_count = []
 
for i in range(len(X)):
    z = zero[i]
    padded = np.r_[False, z, False]
    changes = np.diff(padded.astype(int))
    starts = np.where(changes == 1)[0]
    ends = np.where(changes == -1)[0]

    lengths = ends - starts
    longest_zero_streak.append(lengths.max() if len(lengths) else 0)
    zero_episode_count.append(len(starts))

    valid_adjacent = observed[i, 1:] & observed[i, :-1]
    transitions = (
        (zero[i, 1:] & nonzero[i, :-1]) |
        (nonzero[i, 1:] & zero[i, :-1])
    )
    zero_transition_count.append(
        (transitions & valid_adjacent).sum()
    )

features_compulsory["longest_zero_streak"] = longest_zero_streak
features_compulsory["zero_episode_count"] = zero_episode_count
features_compulsory["zero_transition_count"] = zero_transition_count

# 3. Observed-only daily changes across consecutive calendar days
date_gaps = np.diff(dates.values).astype("timedelta64[D]").astype(int)
consecutive = date_gaps == 1

valid_pairs = (
    observed[:, :-1] &
    observed[:, 1:] &
    consecutive[np.newaxis, :]
)

changes = np.abs(arr[:, 1:] - arr[:, :-1])
changes[~valid_pairs] = np.nan

features_compulsory["valid_change_days"] = valid_pairs.sum(axis=1)

features_compulsory["avg_daily_change"] = np.nanmean(
    changes, axis=1
)
features_compulsory["p95_daily_change"] = np.nanpercentile(
    np.where(np.isfinite(changes), changes, np.nan),
    95,
    axis=1
)

# 4. Observed-only trend and monthly variation
trend_slope = []
monthly_mean_std = []

for i in range(len(X)):
    row = arr[i]
    valid = np.isfinite(row)

    if valid.sum() >= 2:
        t = dates[valid].map(pd.Timestamp.toordinal).to_numpy()
        y = row[valid]
        trend_slope.append(np.polyfit(t - t.mean(), y, 1)[0])
    else:
        trend_slope.append(np.nan)

    monthly = pd.Series(row, index=dates).groupby(
        pd.Index(dates).to_period("M")
    ).mean()

    monthly_mean_std.append(monthly.std())

features_compulsory["trend_slope"] = trend_slope
features_compulsory["monthly_mean_std"] = monthly_mean_std

# 5. Data-quality features from original readings
features_compulsory["missing_percentage"] = (
    X.isna().mean(axis=1) * 100
)
features_compulsory["valid_days"] = X.notna().sum(axis=1)

print("Feature dataframe shape:", features_compulsory.shape)
print("Missing feature values:", features_compulsory.isna().sum().sum())

features_compulsory.head()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import PowerTransformer
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    average_precision_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

# Features proposed for removal in Model 3
noise_features = [
    "cv_consumption",
    "observed_zero_rate",
    "zero_episode_count",
    "zero_transition_count",
    "P10"
]

# Keep only columns that actually exist
noise_features = [
    col for col in noise_features
    if col in X_train.columns
]

# Model 1: Baseline GaussianNB
baseline_nb = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", GaussianNB())
])

# Model 2: Transform all features
transformed_nb = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("transformer", PowerTransformer(method="yeo-johnson")),
    ("model", GaussianNB(priors=[0.5, 0.5]))
])

# Model 3: Transform features after excluding proposed features
selected_columns = [
    col for col in X_train.columns
    if col not in noise_features
]

selected_nb = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("transformer", PowerTransformer(method="yeo-johnson")),
    ("model", GaussianNB(priors=[0.5, 0.5]))
])

# Train each model using its own feature set
baseline_nb.fit(X_train, y_train)

transformed_nb.fit(X_train, y_train)

selected_nb.fit(
    X_train[selected_columns],
    y_train
)

print("Models trained successfully.")
print("Features removed from Model 3:", noise_features)
print("Features retained in Model 3:", len(selected_columns))
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import precision_score, recall_score, f1_score

# Create a validation split from the training data
X_train_sub, X_val, y_train_sub, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    stratify=y_train,
    random_state=42
)

# Select which transformed model to tune
# Change to "selected" to tune the reduced-feature version
model_choice = "transformed"

if model_choice == "transformed":
    tuning_model = transformed_nb
    X_val_used = X_val
    X_train_used = X_train_sub

else:
    tuning_model = selected_nb
    X_val_used = X_val[selected_columns]
    X_train_used = X_train_sub[selected_columns]

# Fit a fresh model only on the smaller training subset
tuning_model.fit(X_train_used, y_train_sub)

val_probs = tuning_model.predict_proba(X_val_used)[:, 1]

threshold_results = []

for threshold in np.arange(0.10, 0.71, 0.05):

    val_pred = (val_probs >= threshold).astype(int)

    threshold_results.append({
        "Threshold": round(threshold, 2),
        "Precision": precision_score(
            y_val, val_pred, zero_division=0
        ),
        "Recall": recall_score(
            y_val, val_pred, zero_division=0
        ),
        "F1-score": f1_score(
            y_val, val_pred, zero_division=0
        )
    })

threshold_df = pd.DataFrame(threshold_results)

print(threshold_df.round(4))

# Choose the threshold with the highest validation F1-score
best_row = threshold_df.loc[threshold_df["F1-score"].idxmax()]
best_threshold = best_row["Threshold"]

print("\nBest threshold:", best_threshold)
print("Validation F1-score:", round(best_row["F1-score"], 4))
```

```python

import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    precision_recall_curve,
    precision_score,
    recall_score,
    f1_score,
    accuracy_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix,
    classification_report
)

# 1. Split the ORIGINAL training data into fitting and validation sets
X_fit, X_val, y_fit, y_val = train_test_split(
    X_train_imputed,
    y_train,
    test_size=0.20,
    stratify=y_train,
    random_state=42
)

# 2. Fit the same best XGBoost pipeline on the fitting portion
# This preserves scaling and resampling inside the pipeline.
threshold_xgb = best_xgb.__class__(
    steps=best_xgb.steps
)

threshold_xgb.fit(X_fit, y_fit)

# 3. Get theft probabilities on the validation set
val_probs = threshold_xgb.predict_proba(X_val)[:, 1]

# 4. Find the threshold that maximizes validation F1
precision, recall, thresholds = precision_recall_curve(
    y_val, val_probs
)

f1_scores = (
    2 * precision[:-1] * recall[:-1]
    / (precision[:-1] + recall[:-1] + 1e-12)
)

best_idx = np.argmax(f1_scores)
best_threshold = thresholds[best_idx]

print("VALIDATION RESULTS")
print("-" * 40)
print(f"Best threshold: {best_threshold:.4f}")
print(f"Validation precision: {precision[best_idx]:.4f}")
print(f"Validation recall: {recall[best_idx]:.4f}")
print(f"Validation F1-score: {f1_scores[best_idx]:.4f}")

# 5. Evaluate on the original, untouched test set
test_probs = threshold_xgb.predict_proba(X_test_imputed)[:, 1]

# Default threshold comparison
test_preds_default = (test_probs >= 0.50).astype(int)

# Validation-selected threshold
test_preds_tuned = (test_probs >= best_threshold).astype(int)

# 6. Compare both thresholds on the test set
def evaluate_threshold(y_true, y_pred, y_probs, label):
    print(f"\n{label}")
    print("-" * 40)
    print(f"Accuracy:  {accuracy_score(y_true, y_pred):.4f}")
    print(f"Precision: {precision_score(y_true, y_pred, zero_division=0):.4f}")
    print(f"Recall:    {recall_score(y_true, y_pred, zero_division=0):.4f}")
    print(f"F1-score:  {f1_score(y_true, y_pred, zero_division=0):.4f}")
    print(f"ROC-AUC:   {roc_auc_score(y_true, y_probs):.4f}")
    print(f"PR-AUC:    {average_precision_score(y_true, y_probs):.4f}")
    print("Confusion matrix:")
    print(confusion_matrix(y_true, y_pred))
    print("\nClassification report:")
    print(classification_report(y_true, y_pred, zero_division=0))

evaluate_threshold(
    y_test, test_preds_default, test_probs,
    "TEST RESULTS — Threshold 0.50"
)

evaluate_threshold(
    y_test, test_preds_tuned, test_probs,
    f"TEST RESULTS — Validation-selected threshold {best_threshold:.4f}"
)
```
